# Strata GPU stages on Kaggle

See `kaggle/README.md` for full setup (GPU accelerator, internet ON, persistence ON, and a `GITHUB_TOKEN` secret). Run the cells in order. Stages 2-4 never call the Anthropic API; stage 2b is optional and off by default.

```python
import os
import subprocess

from kaggle_secrets import UserSecretsClient

token = UserSecretsClient().get_secret("GITHUB_TOKEN")
os.chdir("/kaggle/working")
if not os.path.isdir("strata"):
    subprocess.run(
        ["git", "clone", "--depth", "1",
         f"https://x-access-token:{token}@github.com/JCHETAN26/strata.git"],
        check=True, stdout=subprocess.DEVNULL,
    )
    subprocess.run(["git", "-C", "strata", "remote", "set-url", "origin",
                    "https://github.com/JCHETAN26/strata.git"], check=True)
del token
print("cloned:", os.path.isdir("strata"))
```
